```python
import pandas as pd 
import numpy as np

np.random.seed(42)
n=1200
data={
    "order_ID": range(1001,1001+n),
    "Customer_ID":np.random.randint(101,301,n),
    "Product":np.random.choice(
        ["Laptop","Mobile","Headphones","Keyboard",
         "Mouse","Monitor","Tablet"],n),
     "Category": np.random.choice(
         ["Electronics","Accessories","Computer"],n
     ),
    "city": np.random.choice(
        ["Pune","Mumbai","Nashik","Nagpur","Kolhapur"],n
    ),
    "Quantity":np.random.randint(1,6,n),
    "Price":np.random.randint(500,5001,n),
    "Order_Date":pd.date_range(
        start="2025-01-01",periods=n,freq="D"
    )
}
df =pd.DataFrame(data)
df["Sales"]=df["Quantity"]=df["Price"]
df.to_csv("retail_sales.csv",index=False)
print("Dataset Created Successfully!")
```

```python
df= df.drop_duplicates()
```

```python
#Missing values Handles
df["Price"]=df["Price"].fillna(df["Price"].median())
df["Quantity"]=df["Quantity"].fillna(
    df["Quantity"].median()
)
```

```python
print("Total Rows:",df.shape[0])
print("Total Columns:",df.shape[1])

df.head()
```

```python
total_sales =df["Sales"].sum()
print("Total Sales:",total_sales)
```

```python
#find a total orders
total_orders = df["order_ID"].nunique()

print("Total Orders:", total_orders)

#find total Customers 

total_customers = df["Customer_ID"].nunique()

print("Total Customers:", total_customers)
#find  total average sales
average_sales = df["Sales"].mean()

print("Average Sales:", round(average_sales, 2))
```

```python
product_sales = df.groupby("Product")["Sales"].sum()
print(product_sales.sort_values(ascending=False))
```

```python
#City-wise Sales Analysis
city_sales = df.groupby("city")["Sales"].sum()

print(city_sales.sort_values(ascending=False))
```

```python
#Category-wise Sales Analysis
category_sales = df.groupby("Category")["Sales"].sum()

print(category_sales)
```

```python
#order date convert to datetime
df["Order_Date"] = pd.to_datetime(df["Order_Date"])
```

```python
df["Month"] = df["Order_Date"].dt.month_name()
monthly_sales = df.groupby("Month")["Sales"].sum()

print(monthly_sales)
```

```python
#montly sales graph
plt.figure(figsize=(12, 5))

monthly_sales.plot(kind="bar")

plt.title("Monthly Sales Analysis")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(10, 5))

city_sales.plot(kind="bar", color="orange")

plt.title("City-wise Sales Analysis")
plt.xlabel("City")
plt.ylabel("Total Sales")

plt.show()
```
